```python
conn = sqlite3.connect('../../../field-data-capture.gpkg')
```

```python
tables = ['dic_line_type_artificial',
          'dic_line_type_bedrock',
          'dic_line_type_mass_move',
          'dic_line_type_superficial', 
          'dic_line_type_terrain']

for table in tables:
    df = pd.read_csv(table + '.txt')
    df.description = df.line_type
    df.translation = df.line_type
    df.rename(columns={'line_type': 'code'}, inplace=True)
    df.fid = df.index + 1
    df.to_sql(table, conn, index=False, if_exists='replace')
```
